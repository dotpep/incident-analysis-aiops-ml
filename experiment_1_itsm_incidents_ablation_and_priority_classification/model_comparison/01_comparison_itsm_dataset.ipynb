## Task 5. Check for Data Leakage

**Data leakage** happens when the model receives information that would not legitimately be available at the moment it is supposed to make a prediction, or when a feature effectively gives away the answer.

| Column                | Decision          | Reason                                                                         |
| --------------------- | ----------------- | ------------------------------------------------------------------------------ |
| `Impact`              | Review            | May directly contribute to the priority assignment rule                        |
| `Urgency`             | Review            | May directly contribute to the priority assignment rule                        |
| `CI_Cat`              | Keep              | Describes the affected configuration item                                      |
| `CI_Subcat`           | Keep              | Describes the affected configuration item                                      |
| `Category`            | Keep              | Describes the incident category                                                |
| `Incident_ID`         | Exclude           | Identifier, not a meaningful general predictor                                 |
| `Status`              | Exclude initially | May reflect progress after the incident was opened                             |
| `Open_Time`           | Exclude initially | Could be useful in some scenarios, but is not needed for this first experiment |
| `Resolved_Time`       | Exclude           | Becomes available after resolution                                             |
| `Close_Time`          | Exclude           | Becomes available after closure                                                |
| `Handle_Time_hrs`     | Exclude           | May depend on work completed after the incident began                          |
| `Closure_Code`        | Exclude           | Usually available only after handling the incident                             |
| `No_of_Reassignments` | Exclude initially | Can change during incident handling                                            |


**Data Leakage Check**

Potential data leakage was considered when selecting the features. Incident identifiers were excluded because they do not describe the incident itself. Fields such as resolution time, closure time, handling time, and closure code were excluded because they become available only after the incident has progressed or been resolved.

Impact and urgency were retained as candidate features, but they require special attention because they may be used by the ITSM system to determine priority directly. This could make the prediction task too easy and reduce the value of the experiment. In a future experiment, I would compare results with and without these fields.

The features should represent only information available at the chosen prediction time.


## Task 6. Build the ML Pipeline

**ML Pipeline**

1. Dataset (ITSM incident records in CSV format)

2. Features / Target (X = incident information; y = Priority)

3. Train / Test Split (80% training, 20% testing)

4. Model Selection (Decision Tree Classifier)

5. Training (The model learns patterns from the training set)

6. Prediction (Predict priorities for unseen incidents)

7. Evaluation (Compare predicted and actual priorities using classification metrics)

**ML Pipeline Explanation**

1. **Dataset:** Load the ITSM incident dataset from a CSV file.
2. **Features and Target:** Select the incident characteristics as features (`X`) and `Priority` as the target (`y`).
3. **Train/Test Split:** Divide the data into training and testing sets, using 80% for training and 20% for testing.
4. **Model Selection:** Select a Decision Tree Classifier because the objective is to predict a priority category.
5. **Training:** The future model will learn relationships between the selected features and incident priorities from the training data.
6. **Prediction:** The trained model will predict priorities for incidents in the test set.
7. **Evaluation:** Compare predictions with the actual priorities using accuracy, precision, recall, and F1-score.

Only the dataset preparation, feature selection, data splitting, task identification, model selection, and pipeline design are completed in this laboratory work. Model training, prediction, and evaluation will be performed in a later experiment.
